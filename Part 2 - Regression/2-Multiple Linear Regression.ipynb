# Çoklu Doğrusal REGREASYON
##🟨 2. Multiple Linear Regression (Çoklu Doğrusal Regresyon)
📌 Tanım:

Birden fazla bağımsız değişken ile bir bağımlı değişken arasındaki ilişkiyi modelliyorsun.
📈 Formül:

Y = b0 + b1*X1 + b2*X2 + ... + bn*Xn

🎯 Amaç:

Çoklu faktörleri hesaba katarak Y'yi tahmin etmek.
✅ Örnek:

    X1 = çalışma süresi (saat)

    X2 = önceki sınav ortalaması

    X3 = uyku süresi

    Y = sınav notu

    Amaç: Bu 3 değişkene göre notu tahmin etmek.

 | Model           | Bağımsız Değişken | İlişki Türü       | Kullanım Durumu                     |
| --------------- | ----------------- | ----------------- | ----------------------------------- |
| Simple Linear   | 1                 | Doğrusal (lineer) | X ile Y arasında basit ilişki varsa |
| Multiple Linear | 2+                | Doğrusal (lineer) | Çok faktör varsa                    |
| Polynomial      | 1 (ama X^2, X^3…) | Doğrusal değil    | Eğri (non-linear) bir ilişki varsa  |
        

## Kütüphaneleri Ekleme

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Veriseti Aktarımı Ve Temizleme

```python
dataset = pd.read_csv('50_Startups.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

```python
print(X)
```

```python
print(y)
```

### State Sütununa Kodluyoruz
Encoding categorical data, burada 3. sütunu encodluyoruz

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(), [3])], remainder='passthrough')
X = np.array(ct.fit_transform(X))
print(X)
```

## Verisetini train ve test olarak ayırıyoruz
Splitting the dataset into the training set and test set

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

## Çoklu doğrusal regrasyon modeli ile eğitim
kukla tuzayı ve model oluştururken öğrenmiş olduğumuz 5 maddeyi fonksiyonlar otomatik yapıyor sonuçta bu teknikleri uygulamak iş hayatında zaman kaybına sebeb olmakla birlikte amerikayı tekrar keşfetmeye gerek yoktur kütüphane sınıfları bunları hallediyor, bu durumda veri seti için en uygun modeli seçmek bize kalıyor

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train, y_train) #eğitiyoruz verisetini
#artık söz konusu modeli kullanarak yatırımcılar bu bilgilere dayanarak ne kadar 
#kar edeceklerini tahmin etmek için bu modeli girişimlerine uygulayabilirler
```

## Test setinden tahminler üretmek
Predicting the Test set results
Bu veristetinde birden fazla bağımsız değişken vardır özellikler x de

```python
y_pred = regressor.predict(X_test)
np.set_printoptions(precision=2) #hassasiyet değeri 2
# tahmin edilen karları ve gerçek karları iki vektörünü yan yana göstereceğiz
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

görüldüğü üzere iyi tahminler yapmış diyebiliriz

## Making a single prediction (for example the profit of a startup with R&D Spend = 160000, Administration Spend = 130000, Marketing Spend = 300000 and State = 'California')

```python
print(regressor.predict([[1, 0, 0, 160000, 130000, 300000]]))
```

Dolayısıyla modelimiz, Ar-Ge'ye 160.000, İdari İşlere 130.000 ve Pazarlamaya 300.000 dolar harcayan bir Kaliforniya girişiminin karının 181.566,92 dolar olacağını öngörüyor.

**Önemli not 1:** Özelliklerin değerlerinin hepsinin çift köşeli parantez içinde girildiğine dikkat edin. Bunun nedeni, "öngörme" yönteminin girdilerinin biçimi olarak her zaman 2 boyutlu bir dizi beklemesidir. Ve değerlerimizi çift köşeli parantez içine koymak girdiyi tam olarak 2 boyutlu bir dizi yapar. Basitçe söylemek gerekirse:

$1, 0, 0, 160000, 130000, 300000 \rightarrow \textrm{scalars}$

$[1, 0, 0, 160000, 130000, 300000] \rightarrow \textrm{1D array}$

$[[1, 0, 0, 160000, 130000, 300000]] \rightarrow \textrm{2D array}$

**Önemli not 2:** Ayrıca, "California" durumunun son sütunda bir dize olarak değil, ilk üç sütunda "1, 0, 0" olarak girildiğine dikkat edin. Bunun nedeni, tahmin yönteminin elbette durumun tek sıcak kodlanmış değerlerini beklemesidir ve X özellik matrisinin ikinci satırında gördüğümüz gibi, "California" "1, 0, 0" olarak kodlanmıştır. Ve bu değerleri son üç sütuna değil, ilk üç sütuna dahil etmeye dikkat edin, çünkü kukla değişkenler her zaman ilk sütunlarda oluşturulur.

## Getting the final linear regression equation with the values of the coefficients

```python
print(regressor.coef_)
print(regressor.intercept_)
```

Dolayısıyla çoklu doğrusal regresyon modelimizin denklemi şu şekildedir:

$$
Profit=86.6×Dummy State 1−873×Dummy State 2+786×Dummy State 3+0.773×R&D Spend+0.0329×Administration+0.0366×Marketing Spend+42467.53
$$

**Önemli Not:** Bu katsayıları elde etmek için regressor nesnemizden "coef_" ve "intercept_" niteliklerini çağırdık. Python'daki nitelikler yöntemlerden farklıdır ve genellikle basit bir değer veya bir dizi değer döndürür.
