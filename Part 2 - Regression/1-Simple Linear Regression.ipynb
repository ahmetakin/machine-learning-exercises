# Simple Linear Regression Basit Doğrusal Regresyon

## 🟦 1. Simple Linear Regression (Basit Doğrusal Regresyon)
📌 Tanım:

Sadece bir bağımsız değişken (X) ile bir bağımlı değişken (Y) arasındaki ilişkiyi modelleyen regresyon türüdür.
📈 Formül:

Y = b0 + b1 * X

    b0: Y'nin X=0 olduğu durumda tahmini değeri (intercept - kesişim noktası)

    b1: X değişkeninin eğimi, yani X arttığında Y ne kadar değişir

🎯 Amaç:

X arttıkça Y nasıl değişiyor? Tahmin edebilmek.
✅ Örnek:

    X = çalışma süresi (saat)

    Y = sınavdan alınan not

    Amaç: 5 saat çalışan birinin notunu tahmin etmek.

 | Model           | Bağımsız Değişken | İlişki Türü       | Kullanım Durumu                     |
| --------------- | ----------------- | ----------------- | ----------------------------------- |
| Simple Linear   | 1                 | Doğrusal (lineer) | X ile Y arasında basit ilişki varsa |
| Multiple Linear | 2+                | Doğrusal (lineer) | Çok faktör varsa                    |
| Polynomial      | 1 (ama X^2, X^3…) | Doğrusal değil    | Eğri (non-linear) bir ilişki varsa  |
   

## Burada Ne ile çalışacağız
Burada daha önceteden temizlenmiş bir maaş verisi üzerinden çalışacağız veri içerisinde deneyim yılı ve aldıkları maaşı içeren bir verisetimiz olacak

## Kütüphaneleri Ekleme

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.model_selection import train_test_split
```

## Verisetini ekleme ve eğitim,test olarak bölme

```python
dataset = pd.read_csv('Salary_Data.csv')
X = dataset.iloc[:, :-1].values #bagımsız değişken sütunu
y = dataset.iloc[:, -1].values #bağımlı değişken sütunu

x_train,x_test,y_train,y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
## Linear Regrasyon modeli oluğturup verileri içine aktarıyoruz
```

```python
from sklearn.linear_model import LinearRegression
regressor =  LinearRegression() #model oluşturduk
regressor.fit(x_train,y_train)
```

## Test setlerini Tahmin etmek

```python
y_pred = regressor.predict(x_test) #
```

## Eğitim veriseti sonuçlarını görselleştirme
regresyon çizgisi tahmin edilen sonuçları verir burada 

```python
plt.scatter(x_train,y_train, color = 'red')
plt.plot(x_train, regressor.predict(x_train),color = 'blue')
plt.title('Salary vs Experience (Training set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

## Test verisetini sonuçlarını görselleştirme

```python
plt.scatter(x_test,y_test, color = 'red')
plt.plot(x_train, regressor.predict(x_train),color = 'blue')
plt.title('Salary vs Experience (Test set)')
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.show()
```

Burada düzgün bir çizgi oluşturma sebebi verisetindeki verileri doğrusal bir şekilde ilerlemesidir her veriseti için böyle anlamlı sonuçlar olmayacaktır bu sadece basit  bir örnek

## Şimdi Tahmin yapmayı görelim örneğin 12 yıllık deneyimi olacak birisi için

```python
print(regressor.predict([[12]]))
```

Görüldüğü üzere 12 yıllık deneyimi olan bir kişinin alacağı maaşı hesapladık

**Önemli:** Özelliğin değerinin (12 yıl) çift köşeli parantez içinde girildiğine dikkat edin. Bunun nedeni, "öngörme" yönteminin girdilerinin biçimi olarak her zaman 2 boyutlu bir dizi beklemesidir. Ve 12'yi çift köşeli parantez içine koymak girdiyi tam olarak 2 boyutlu bir dizi yapar. Basitçe söylemek gerekirse:

$12 \rightarrow \textrm{scalar}$

$[12] \rightarrow \textrm{1D array}$

$[[12]] \rightarrow \textrm{2D array}$

## Katsayıların değerleri ile son doğrusal regresyon denklemini elde etmek

```python
print(regressor.coef_)
print(regressor.intercept_)
```

Bu nedenle, basit doğrusal regresyon modelimizin denklemi şudur:

$$\textrm{Salary} = 9345.94 \times \textrm{YearsExperience} + 26816.19$$
