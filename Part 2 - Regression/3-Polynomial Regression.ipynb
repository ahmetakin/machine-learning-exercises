## 🟩 3. Polynomial Regression (Polinom Regresyon)
📌 Tanım:

Veri noktaları düz bir doğruyla değil, kıvrımlı (non-linear) bir yapı ile daha iyi ifade ediliyorsa kullanılır.
📈 Formül:

Y = b0 + b1*X + b2*X^2 + b3*X^3 + ... + bn*X^n

🎯 Amaç:

X ve Y arasında doğrusal olmayan (non-linear) ilişkileri modellemek.
✅ Örnek:

    X = araba hızı

    Y = yakıt tüketimi

    Hız arttıkça tüketim önce azalır sonra artar → bu kıvrımlı ilişki için polinom gerekir.
şimdi elimizde mülakata gelen bir kişinin eski şirketin pozisyonlara göre maaşı olduğunu varsayalım


 | Model           | Bağımsız Değişken | İlişki Türü       | Kullanım Durumu                     |
| --------------- | ----------------- | ----------------- | ----------------------------------- |
| Simple Linear   | 1                 | Doğrusal (lineer) | X ile Y arasında basit ilişki varsa |
| Multiple Linear | 2+                | Doğrusal (lineer) | Çok faktör varsa                    |
| Polynomial      | 1 (ama X^2, X^3…) | Doğrusal değil    | Eğri (non-linear) bir ilişki varsa  |

## İmporting the Libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## importing the dataset

```python
dataset = pd.read_csv('Position_Salaries.csv')
X = dataset.iloc[:, 1:-1].values # 2 SÜTUNU ALDIM ORTAKİ LEVEL
y = dataset.iloc[:, -1].values
```

## Training the Linear Regression model on the whole dataset 
önce tüm verinin linear regressoyununu alalım

```python
from sklearn.linear_model import LinearRegression
lin_reg = LinearRegression()
lin_reg.fit(X,y)
```

## Training the Polynomial Regression model on the whole dataset 
Şimdi polinum regrasyonunu alalım tüm datanın

```python
from sklearn.preprocessing import PolynomialFeatures
poly_reg = PolynomialFeatures(degree = 4) #degree formüldeki n değeri üssü biz belirliyoruz 3-4 deneyelim
X_poly = poly_reg.fit_transform(X)
lin_reg_2 = LinearRegression()
lin_reg_2.fit(X_poly,y)
```

## Visualising the Linear Regression results
linear regrasyonu görselleştirelim

```python
plt.scatter(X,y, color= 'red')
plt.plot(X,lin_reg.predict(X), color = 'blue') #tahminleri içeren regrasyon çizgisini
plt.title('Truth or Bluff (Linear Regression)')
plt.xlabel('İş pozisyonu')
plt.ylabel('Maaş')
plt.show()
#dikkat edilirse regrasyon çok anlamız geliyor çizgi
```

## Visualising the Polynomial Regression results
polinom regrasyonunu sonuçları görselleştirelim

```python
plt.scatter(X, y, color = 'red')
plt.plot(X, lin_reg_2.predict(X_poly), color = 'blue')
plt.title('Truth or Bluff (Polynomial Regression)')
plt.xlabel('is seviyesi')
plt.ylabel('Maaş')
plt.show()
#dikkat edilirse daha anlamlı regrasyon tahmin çizgisi çizecek
#degree değiştikçe daha iyi hesaplar yapıcak çizgi düzelecek
```

## Visualising the Polynomial Regression results (for higher resolution and smoother curve)
Polinom Regresyon sonuçlarının görselleştirilmesi (daha yüksek çözünürlük ve daha düzgün eğri için)

```python
X_grid = np.arange(min(X), max(X), 0.1)
X_grid = X_grid.reshape((len(X_grid), 1))
plt.scatter(X, y, color = 'red')
plt.plot(X_grid, lin_reg_2.predict(poly_reg.fit_transform(X_grid)), color = 'blue')
plt.title('Truth or Bluff (Polynomial Regression)')
plt.xlabel('İş seviyesi')
plt.ylabel('Maaş')
plt.show()
```

## Doğrusal Regresyon ile yeni bir sonucu tahmin etme
Predicting a new result with Linear Regression

```python
lin_reg.predict([[6.5]])
```

## Polinom Regresyonu ile yeni bir sonucu tahmin etme
Predicting a new result with Polynomial Regression

daha doğru tahminde bulunacak

```python
lin_reg_2.predict(poly_reg.fit_transform([[6.5]]))
```
