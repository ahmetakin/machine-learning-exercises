---

## 📌 1. PCA Nedir? Ne İşe Yarar?

PCA, **çok sayıda özelliğe (feature)** sahip bir veri kümesini, **daha az sayıda ama bilgi açısından en zengin** olan yeni özelliklere dönüştürür.

Amaç:
🔻 **Boyutları azaltmak**,
🔻 **Bilgiyi (varyansı) kaybetmeden veriyi sadeleştirmek**,
🔻 Görselleştirmeyi kolaylaştırmak,
🔻 Model eğitimini hızlandırmak ve overfitting’i azaltmak.

---

## 📊 2. Gerçek Hayattan Basit Örnek

Hayal et:

* Elinde 1000 kişinin **boyu (cm)** ve **kilosu (kg)** var.
* Bunları iki boyutta (X=boy, Y=kilo) bir grafikte gösteriyorsun.
* Şimdi, bu iki veriden yeni bir eksen oluşturalım:
  📈 **“vücut büyüklüğü”** gibi bir kavram. Boy ve kilo birlikte değişiyor çünkü.

PCA burada şunu yapar:

> "Bu iki değişkenden (boy ve kilo), bilgiyi en iyi taşıyan **yeni bir eksen** oluşturayım. Böylece artık veriyi **tek bir sayıyla** (yeni eksen üzerindeki konum) temsil edebilirim."

---

## 🎯 3. PCA'nın Aradığı Şey: **En Yüksek Varyans Yönü**

PCA, verideki en büyük **farklılığı (varyansı)** taşıyan yönleri bulur.

🔹 1. bileşen (PC1): Varyansın en fazla olduğu yön
🔹 2. bileşen (PC2): PC1’e dik olan ve hala bilgi taşıyan ikinci yön
🔹 ... ve böyle devam eder

Bu bileşenleri bulmak için şu işlemleri yapar:

* Veriyi merkeze çeker (ortalama sıfırlama)
* Kovaryans matrisini hesaplar
* Bu matristen **özvektör ve özdeğerleri** bulur
* En büyük özdeğere sahip vektör = **en çok bilgiyi taşıyan yön**

> **Matematiksel gibi görünebilir ama özü şudur:**
> PCA = Enformasyonun en yoğun olduğu yönleri bulur ve bu yönlerde veriyi yeniden çizer.

---

## 🧠 4. Görselle Anlatım (Kareli Kağıt Üzerinde)

Hayal et ki verin şu şekilde dağılmış:

```
|
|          o      o
|       o     o
|    o     o
| o     o
|________________________
```

* Bu noktalar diyelim ki boy-kilo verisi.
* PCA der ki: “Ben bu dağılımın uzandığı yönü **diyagonal** olarak çizeyim (en çok bilgi burada).”
* Yeni eksen **diyagonal** olur, yani veriyi o eksen boyunca ifade edersin.
* Böylece veriyi tek boyutta bile özetleyebilirsin!

---

## 🧪 5. PCA Uygulaması – Özet Adımlar

Diyelim ki elimizde 3 özellikli veri var:

```
[boy, kilo, yaş]
```

### PCA Uygulaması:

1. **Standartlaştır** veriyi (ortalama=0, std=1)
2. **Kovaryans matrisini** hesapla
3. Özdeğer & özvektörleri bul
4. En büyük özdeğere sahip vektörleri al (örneğin 2 tanesi %95 bilgiyi açıklıyorsa)
5. Bu vektörlerle veriyi **yeni uzaya** yansıt (transform)

Artık veri şuna döner:

```
[PC1, PC2] → 2 boyuta indirilmiş hali
```

---

## 💡 6. Ne Zaman Kullanılır?

* Özellik sayısı fazla ama örnek sayısı azsa (overfitting riski)
* Görselleştirme isteniyorsa (örneğin 10D → 2D)
* Özellikler arasında korelasyon varsa
* Modelin hızlandırılması gerekiyorsa

---

## 🔁 PCA’nın Zayıf Noktaları

* Sadece **lineer ilişkileri** yakalar (non-lineer değil)
* Özelliklerin yorumlanabilirliğini azaltabilir
* Etiket bilgisi **kullanmaz** → sınıfları ayırmakta yetersiz kalabilir (LDA bu yüzden devreye girer)

---

## ✅ Özetle PCA Şunu Yapar:

🔹 Çok boyutlu veriyi, en bilgi taşıyan yeni boyutlara **indirger**
🔹 Etiket gerekmez (unsupervised)
🔹 Görselleştirme ve hız için idealdir
🔹 Yeni özellikler = “Principal Components”

Çok iyi bir soru! PCA (Principal Component Analysis) aslında bir model değil, bir **ön işleme / özellik çıkarımı (feature extraction)** yöntemidir. Ama doğru kullanıldığında **çok güçlü** modellerin temelini oluşturur.

## 🎯 PCA ile Ne İşe Yarar?

### 🔹 1. **Boyut Azaltma (Dimensionality Reduction)**

* 100 özellikli bir veriyi → 2 ya da 3 bileşene indirerek:

  * **Modeli hızlandırırsın**
  * **Overfitting'i azaltırsın**
  * **Veriyi görselleştirebilirsin**

### 🔹 2. **Gürültü Azaltma (Noise Filtering)**

* PCA varyansı yüksek yönleri alır. Gürültü (noise) genellikle varyansı düşük boyutlardadır.
  → PCA ile **gürültüyü elersin**, daha temiz veriyle çalışırsın.

### 🔹 3. **Korelasyon Problemini Çözme**

* Özelliklerin (feature'ların) birbiriyle **aşırı ilişkili** olduğu durumlarda (multicollinearity), PCA yeni bağımsız bileşenler üretir.
  → Böylece daha **sağlam modeller** kurabilirsin.

---

## 🤖 PCA Sonrası Hangi Modeller Kullanılır?

PCA, veriyi dönüştürür. Sonrasında her türlü **denetimli öğrenme (supervised learning)** algoritmasını kullanabilirsin:

### 🔸 Sınıflandırma (Classification)

* PCA → Lojistik Regresyon
* PCA → Naive Bayes
* PCA → SVM (çok popüler kombinasyon!)
* PCA → KNN
* PCA → Random Forest / XGBoost

### 🔸 Regresyon (Regression)

* PCA → Linear Regression (özellikle çoklu korelasyon varsa)
* PCA → Ridge / Lasso Regression

### 🔸 Kümeleme (Clustering)

* PCA → K-Means
* PCA → DBSCAN
* PCA → Hierarchical Clustering

### 🔸 Görselleştirme (Visualization)

* Özellikle 3’ten fazla özelliğe sahip verilerde, PCA sayesinde 2D veya 3D grafiklerde kümeleri veya ayrımları gösterebilirsin.
  (Örneğin: PCA + t-SNE ile veri analizi)

---

## 💡 PCA Hangi Durumlarda Gerçekten İşine Yarar?

| Durum                           | PCA İşe Yarar mı? | Neden?                               |
| ------------------------------- | ----------------- | ------------------------------------ |
| Çok fazla feature varsa         | ✅                 | Model karmaşıklığını azaltır         |
| Görselleştirme istiyorsan       | ✅                 | 2D/3D’ye indirger                    |
| Özellikler korelasyonluysa      | ✅                 | Bağımsız bileşenlere çevirir         |
| Etiketli veri yoksa             | ✅                 | PCA **unsupervised**’dır             |
| Non-lineer yapı varsa           | ❌                 | PCA sadece lineer ilişkileri yakalar |
| Özellikler zaten iyi ayrılmışsa | ❌                 | PCA bazen ayrımı bozabilir           |

## Importing the libraries and Dataset

````python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
````

````python
dataset = pd.read_csv('Wine.csv')
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
````

## Splitting the dataset into the Training set and Test set

````python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
````

## Feature Scaling

````python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
````

## Applying PCA

````python
from sklearn.decomposition import PCA 
#pca sınıflandırma modelincen önce uygulanır unutma
pca = PCA(n_components = 2) #2 ye düşürüyoruz bu sayıya 2 boyut seçtik eğer düzgün sonuç göremez isek rakamı yükseltebiliriz
X_train = pca.fit_transform(X_train)
X_test = pca.transform(X_test)
````

## Training the Logistic Regression model on the Training set

````python
from sklearn.linear_model import LogisticRegression
#ille logistic kullanmana gerek yok istediğin modeli uygulayabilirsin
classifier = LogisticRegression(random_state = 0)
classifier.fit(X_train, y_train)
````

## Making the Confusion Matrix

````python
from sklearn.metrics import confusion_matrix, accuracy_score
y_pred = classifier.predict(X_test)
cm = confusion_matrix(y_test, y_pred)
print(cm)
accuracy_score(y_test, y_pred)
````

## Visualising the Training set results

````python
from matplotlib.colors import ListedColormap
X_set, y_set = X_train, y_train
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = 0.01))
plt.contourf(X1, X2, classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green', 'blue')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = ListedColormap(('red', 'green', 'blue'))(i), label = j)
plt.title('Logistic Regression (Training set)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.show()
````

## Visualising the Test set results

````python
from matplotlib.colors import ListedColormap
X_set, y_set = X_test, y_test
X1, X2 = np.meshgrid(np.arange(start = X_set[:, 0].min() - 1, stop = X_set[:, 0].max() + 1, step = 0.01),
                     np.arange(start = X_set[:, 1].min() - 1, stop = X_set[:, 1].max() + 1, step = 0.01))
plt.contourf(X1, X2, classifier.predict(np.array([X1.ravel(), X2.ravel()]).T).reshape(X1.shape),
             alpha = 0.75, cmap = ListedColormap(('red', 'green', 'blue')))
plt.xlim(X1.min(), X1.max())
plt.ylim(X2.min(), X2.max())
for i, j in enumerate(np.unique(y_set)):
    plt.scatter(X_set[y_set == j, 0], X_set[y_set == j, 1],
                c = ListedColormap(('red', 'green', 'blue'))(i), label = j)
plt.title('Logistic Regression (Test set)')
plt.xlabel('PC1')
plt.ylabel('PC2')
plt.legend()
plt.show()
````
