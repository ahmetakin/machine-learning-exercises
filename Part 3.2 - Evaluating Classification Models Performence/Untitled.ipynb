Bu bölümde öğreneceğin konular özellikle **kurumsal dünyada** (finans, pazarlama, kredi skorlama gibi) modellerin performansını daha iyi analiz etmek için kullanılır.

# 🔷 227. Video: **Understanding CAP Curves**

## 🎯 CAP (Cumulative Accuracy Profile) Nedir?

CAP eğrisi, sınıflandırma modelinin **ne kadar başarılı olduğunu kümülatif olarak** gösteren bir grafiktir.

### 📉 Nasıl çalışır?

* Tüm gözlemleri **modelin tahmin olasılığına göre sıralarsın**
* En yüksek olasılıktan en düşük olasılığa doğru ilerlersin
* X ekseni: Gözlem yüzdesi (0% → 100%)
* Y ekseni: Doğru sınıf tahmini yapılan gözlem yüzdesi (kümülatif)

---

### 📊 CAP Grafiğinde 3 Eğri Olur:

| Eğri                   | Anlamı                                                |
| ---------------------- | ----------------------------------------------------- |
| **Random Model (düz)** | Model yok, rastgele seçim yapıyor                     |
| **Perfect Model**      | Mükemmel model → %x gözlemle %x doğru pozitif yakalar |
| **Gerçek Model (CAP)** | Sizin modelin performans eğrisi                       |

📌 Modelin CAP eğrisi ne kadar **"perfect" eğriye yakınsa**, model o kadar iyidir.

---

# 🔷 228. Video: **Accuracy Ratio (AR) – CAP ile Model Kıyaslama**

## 🎯 Accuracy Ratio (AR) Nedir?

CAP eğrisi ile hesaplanan **tek bir sayısal skor**.
Bu skor, modelin rastgele tahmin yerine **gerçek sınıfları ne kadar iyi ayrıştırabildiğini** gösterir.

### 📐 Hesap Formülü:

```
AR = (A_model - A_random) / (A_perfect - A_random)
```

* `A_model`: CAP eğrisi altında kalan alan (senin modelin)
* `A_random`: Rastgele modelin altındaki alan (düz çizgi)
* `A_perfect`: Mükemmel modelin altındaki alan

### 🎯 Yorum:

| AR Skoru | Anlamı                      |
| -------- | --------------------------- |
| 1.0      | Mükemmel model              |
| 0.0      | Rastgele model kadar kötü   |
| < 0      | Rastgele modelden bile kötü |
| 0.6–0.9  | Gerçek hayatta iyi model    |

---

## 📈 Neden Kullanılır?

* **ROC eğrisine benzese de**, daha sezgisel bir değerlendirme sunar
* Özellikle **kredi riski, churn analizi, fraud detection** gibi alanlarda çok sık tercih edilir
* Kurumsal firmalar model kabulünde **AR oranı > 0.6** gibi eşiklere göre karar verir

---

| ----------------------- | ------------------------------------------------------ |
| **CAP Curve**           | Modelin kümülatif başarısını görsel olarak gösterir    |
| **Accuracy Ratio (AR)** | CAP eğrisi altında kalan alanla ölçülen skor           |
| **Ne işe yarar?**       | Modelin rastgeleye göre ne kadar iyi olduğunu gösterir |
