# 📌 Association Rule Learning Nedir?

Association Rule Learning, **veri içerisindeki öğeler (özellikler, ürünler, olaylar) arasındaki ilişkileri/kombinasyonları** bulmak için kullanılır.

### En bilinen örnek:

> "Market sepeti analizi" (Market Basket Analysis)

> ❓“Müşteri ekmek aldıysa, yanında genellikle ne alıyor?”
> ❓“Bir ürünü alan müşteriler genellikle hangi ürünleri de alır?”

Yani:

* **Eş zamanlı görülen öğeler**
* **Birlikte satın alınan ürünler**
* **Sebep–sonuç olabilecek kural desenleri**

---

## 🧾 Kurallar Şu Formatta Olur:

```
A ⇒ B
```

> "Eğer A varsa, B olma olasılığı yüksektir"

Örnek:

```
{Ekmek, Süt} ⇒ {Yumurta}
```

---

## 📊 Kullanılan Metrikler:

| Metrik         | Açıklama                                                                                    |
| -------------- | ------------------------------------------------------------------------------------------- |
| **Support**    | A ve B’nin birlikte görülme oranı (destek)                                                  |
| **Confidence** | A olduktan sonra B’nin de görülme olasılığı (güven)                                         |
| **Lift**       | A ve B'nin birlikte görülmesi, B'nin bağımsız görülmesine göre ne kadar güçlü? (anlamlılık) |

---

## 📌 Apriori Algoritması Nedir?

**Apriori**, Association Rule Learning’de en çok bilinen algoritmadır.

> Amaç: “En sık birlikte görülen öğe kümelerini bul, sonra bu kümelerden anlamlı kurallar türet.”

---

### 🔁 Nasıl Çalışır?

1. **Destek değeri eşik (min support)** belirlenir
2. Bu eşiği geçen öğeler (tekli, ikili, üçlü vb.) bulunur
3. Bu sık öğe kümelerinden **A ⇒ B** şeklinde kurallar üretilir
4. Kuralların **confidence** ve **lift** değerleri hesaplanır
5. Anlamlı olanlar filtrelenir

---

### 🎯 Neden "Apriori"?

Çünkü algoritma "Apriori özelliği"ni kullanır:

> "Eğer bir öğe grubu nadirse, onun alt kümeleri de nadirdir"
> Bu sayede **gereksiz kombinasyonlar elenir**, işlem hızlanır

---

## ✅ Avantajları:

| Özellik                   | Açıklama                                               |
| ------------------------- | ------------------------------------------------------ |
| Yorumlaması kolay         | “Eğer bu varsa, bu da olur” tarzı net kurallar üretir  |
| Öneri sistemlerinde güçlü | Ürün, içerik, kullanıcı alışkanlıkları için kullanılır |
| Sıklık tabanlı            | Güçlü korelasyonları kolay yakalar                     |

---

## ⚠️ Dezavantajları:

| Sınırlama                            | Açıklama                                             |
| ------------------------------------ | ---------------------------------------------------- |
| Büyük veriyle yavaş                  | Kombinasyonlar çok fazla olabilir                    |
| Min support/confidence seçimi önemli | Yanlış eşikler ya çok az ya da alakasız kural üretir |
| Sadece varlık yokluk bakar           | Nicel değerleri (örneğin miktar) direkt işlemez      |

---

## 💻 Python'da Apriori Örneği (mlxtend kütüphanesi ile):

```python
from mlxtend.frequent_patterns import apriori, association_rules

# 1. Sık öğe kümelerini bul
frequent_itemsets = apriori(df, min_support=0.01, use_colnames=True)

# 2. Kuralları üret
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.0)

# 3. Güçlü kuralları filtrele
strong_rules = rules[(rules['confidence'] > 0.6) & (rules['lift'] > 1.2)]
```

````python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
````

````python
dataset = pd.read_csv('Market_Basket_Optimisation.csv', header = None) # headeri yok
transactions = []
for i in range(0, 7501):
  transactions.append([str(dataset.values[i,j]) for j in range(0, 20)])
````

## Training the Apriori model on the dataset

````python
from apyori import apriori
rules = apriori(
            transactions = transactions, 
            min_support = 0.003, # en az 3 ürün alımı.  * en. çok ürün alımı  / veri satır. sayısı
            min_confidence = 0.2,   # rastgele. veriye göre denenerek seçilir
            min_lift = 3, # 3 olsun dedi
            min_length = 2, 
            max_length = 2) #şimdi bir ürün alana 2. bedava vereceğimiz için 2 yaptı min max ı
````

Bu çok önemli bir soru ve **Apriori algoritmasında başarıyı belirleyen en kritik yer tam olarak burası**:
Yani `min_support`, `min_confidence`, `min_lift`, `min_length`, `max_length` gibi **eşik (threshold)** değerlerinin **neden seçildiği ve nasıl değiştiği** konusu.

Şimdi sana bu parametrelerin **ne işe yaradığını**, **neden böyle verildiğini** ve **nasıl ayarlanması gerektiğini** açıklayayım:

---

## 🔍 Kod Parçası:

```python
rules = apriori(
    transactions = transactions, 
    min_support = 0.003,
    min_confidence = 0.2,  
    min_lift = 3,

```

---

## 1️⃣ `min_support = 0.003`

### 📌 Ne demek?

* En az %0.3 destek oranına sahip öğe kombinasyonlarını dikkate al.

### 🎯 Neden böyle?

* Diyelim veri setinde 10.000 işlem var.
* 0.003 destek = **en az 30 işlemde görülen kombinasyonlar**
* Çok nadir görülen kombinasyonlar göz ardı edilir

### 🧠 Not:

* Büyük veri setinde küçük değer mantıklıdır
* Küçük veri setinde daha yüksek destek gerekir (`0.05`–`0.1` gibi)

---

## 2️⃣ `min_confidence = 0.2`

### 📌 Ne demek?

* A ⇒ B kuralının en az %20 güvenilirliği (doğruluğu) olmalı

### 🎯 Neden böyle?

* `P(B | A)` ≥ 0.2 olmalı → A varsa B’nin de gelme olasılığı %20’nin altında olmasın

### 🧠 Not:

* Düşük confidence: zayıf ilişki
* Yüksek confidence: daha anlamlı ama az sayıda kural çıkarır
* Genelde `0.5`–`0.8` arası seçilir
* Ama burada **daha çok kural çıkarılsın diye 0.2 verilmiş**

---

## 3️⃣ `min_lift = 3`

### 📌 Ne demek?

* Lift ≥ 3 olacak → A ve B birlikte olması, B’nin bağımsız oluşundan **en az 3 kat daha olası**

### 🎯 Neden böyle?

* Lift = 1 → rastgele ilişki
* Lift > 1 → olumlu ilişki
* Lift ≥ 3 → **kuvvetli ilişki** demektir (süpermarket verisinde sık kullanılır)

### 🧠 Not:

* `lift > 1.2` → anlamlı
* `lift > 2` → güçlü
* `lift > 3` → çok güçlü
* Bu eşik, çıkarılan kuralların **gerçekten güvenilir** olmasını sağlar

---

## 4️⃣ `min_length = 2`, `max_length = 2`

### 📌 Ne demek?

* Kurallar sadece **2 öğe** içersin (`A ⇒ B` gibi)

### 🎯 Neden böyle?

* Basit, okunabilir ve pratik kurallar istenmiş
* Örnek: `{süt} ⇒ {ekmek}`

### 🧠 Not:

* Eğer `min_length=2`, `max_length=3` dersen
  → 2’li ve 3’lü kuralları da alırsın: `{süt, tereyağı} ⇒ {ekmek}` gibi
* Daha büyük kombinasyonlar = daha karmaşık ama bazen daha ilginç sonuçlar

---

## 🧠 Sonuç: Bu Eşikler “Sabit” Değil

Bunlar:

✅ Veri büyüklüğüne
✅ İşin doğasına (market verisi mi, tıklama mı, sağlık mı)
✅ Beklenen sonuçların yoğunluğuna göre
**değiştirilebilir**

## 📊 Örnek Alternatif Değer Setleri:

| Kullanım Amacı             | min\_support | min\_confidence | min\_lift | Açıklama                                    |
| -------------------------- | ------------ | --------------- | --------- | ------------------------------------------- |
| Market verisi (büyük veri) | 0.002–0.005  | 0.2–0.5         | 2–3       | Çok işlem var, sık gelen kombinasyon aranır |
| Küçük veri seti            | 0.05–0.1     | 0.5–0.8         | 1.2+      | Az veri olduğunda eşikler yükseltilir       |
| Yoğun öneri çıkarma        | 0.001        | 0.1             | 1         | Daha fazla ama zayıf kural çıkarılır        |

---

## 🎓 Özetle:

| Parametre        | Ne İşe Yarar?                            |
| ---------------- | ---------------------------------------- |
| `min_support`    | En az kaç kez görünmeli? (frekans eşiği) |

| `min_lift`       | Rastgeleliğe göre ne kadar güçlü ilişki? |
| `min_length`     | En kısa kaç öğeli kural?                 |
| `max_length`     | En uzun kaç öğeli kural?                 |

### Putting the results well organised into a Pandas DataFrame

````python
def inspect(results):
    lhs         = [tuple(result[2][0][0])[0] for result in results]
    rhs         = [tuple(result[2][0][1])[0] for result in results]
    supports    = [result[1] for result in results]
    confidences = [result[2][0][2] for result in results]
    lifts       = [result[2][0][3] for result in results]
    return list(zip(lhs, rhs, supports, confidences, lifts))
resultsinDataFrame = pd.DataFrame(inspect(results), columns = ['Left Hand Side', 'Right Hand Side', 'Support', 'Confidence', 'Lift'])
````
