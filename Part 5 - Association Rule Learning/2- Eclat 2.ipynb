```python
from pyECLAT import ECLAT
import pandas as pd
 
# Load the dataset
dataset = pd.read_csv('Market_Basket_Optimisation.csv', header=None)
 
# Convert the dataset to string
transactions = dataset.applymap(str)
 
# Instantiate Eclat
eclat_instance = ECLAT(data=transactions)
 
# Define the minimum support
min_support = 0.003
 
# Get frequent itemsets
index, support = eclat_instance.fit(min_support=min_support, verbose=False)
 
# Create a DataFrame to store the results
results = pd.DataFrame({'Itemset': index, 'Support': support})
results
```
