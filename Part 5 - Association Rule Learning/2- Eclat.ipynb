# 🟦 **ECLAT (Equivalence Class Clustering and bottom-up Lattice Traversal)**

## – Apriori’ye göre daha hızlı, destek (support) odaklı bir algoritma

---

## 📌 Temel Fark: Apriori vs ECLAT

| Özellik          | **Apriori**                        | **ECLAT**                                 |
| ---------------- | ---------------------------------- | ----------------------------------------- |
| Tarama yönü      | Yatay tarama (transaction list)    | Dikey tarama (item list)                  |
| Veri yapısı      | Liste liste gider (`transactions`) | Item-tablosu (örneğin sütuna göre ID’ler) |
| Hedef            | Sık itemset'leri bulmak            | Aynı                                      |
| Kural üretimi    | Ekstra aşama gerekir               | Genelde sadece **sık itemset çıkarır**    |
| Hız / performans | Büyük veride yavaş olabilir        | Daha az bellek kullanır, daha hızlı       |

---

## 🎯 ECLAT Ne Yapar?

1. Tüm ürünler için hangi işlemlerde (ID'lerde) yer aldığını çıkartır
   → Örnek:

   ```
   Süt:     [1, 2, 3, 5]  
   Ekmek:   [2, 3, 5, 6]
   ```

2. Bu listeleri **kesiştirerek** hangi kombinasyonların kaç kez geçtiğini bulur

   * `{Süt, Ekmek}` → `[2, 3, 5]` → support = 3 / toplam işlem sayısı

3. Belirli `min_support` üzerindeki **sık itemset’leri çıkarır**

> NOT: ECLAT, sadece "hangi kombinasyonlar sık görülüyor"u verir, **A⇒B** tarzı kuralları doğrudan vermez (onları ayrıca hesaplaman gerekir).

---

## 🧮 Teknik Olarak Nasıl Çalışır?

Örnek veri:

```
Transaction 1: Süt, Ekmek  
Transaction 2: Süt, Yumurta  
Transaction 3: Süt, Ekmek  
Transaction 4: Ekmek, Yumurta
```

ECLAT şunu yapar:

* Süt → {1, 2, 3}
* Ekmek → {1, 3, 4}
* Kesişim: {1, 3} → support = 2

→ `{Süt, Ekmek}` support = 2 / 4 = 0.5

---

## ✅ Avantajları:

| Avantaj               | Açıklama                            |
| --------------------- | ----------------------------------- |
| Daha hızlıdır         | Özellikle büyük veri setlerinde     |
| Daha az bellek harcar | Dikey temsil (bit tabanlı olabilir) |
| Sade ve etkili        | Hızlı sık itemset çıkarımı yapar    |

---

## ⚠️ Sınırlamaları:

| Dezavantaj                 | Açıklama                                     |
| -------------------------- | -------------------------------------------- |
| Kural çıkarımı yok         | A⇒B gibi kurallar için ekstra işlem gerekir  |
| Büyük itemset birleşimleri | Çok sayıda öğe olduğunda yine yavaşlayabilir |
| Support’a odaklıdır        | Confidence ve lift gibi metrikleri içermez   |

## 💻 Python Kullanımı (genelde özel kütüphaneler gerekir)

Scikit-learn veya `mlxtend` gibi popüler kütüphaneler ECLAT içermez.
Ama bazı özel örneklerde ECLAT şöyle benzetilerek uygulanabilir:

```python
# Dikey gösterim benzeri
item_id_dict = {
    "süt": {1, 2, 3},
    "ekmek": {1, 3, 4},
    "yumurta": {2, 4}
}

# Kesişim ile destek hesaplama
süt_ve_ekmek = item_id_dict["süt"] & item_id_dict["ekmek"]
support = len(süt_ve_ekmek) / toplam_trans_sayısı
```

Alternatif olarak bazı ECLAT implementasyonları R, Java veya özel Python script'lerde vardır (örnek: PyECLAT).

````python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
````

````python
dataset = pd.read_csv('Market_Basket_Optimisation.csv', header = None)
transactions = []
for i in range(0, 7501):
  transactions.append([str(dataset.values[i,j]) for j in range(0, 20)])
````

## Training the Eclat model on the dataset

````python
from apyori import apriori
rules = apriori(
    transactions = transactions, 
    min_support = 0.003, 
    min_confidence = 0.2, 
    min_lift = 3, 
    min_length = 2, 
    max_length = 2)
````

````python
results = list(rules)
````

````python
results
````

````python
def inspect(results):
    lhs         = [tuple(result[2][0][0])[0] for result in results]
    rhs         = [tuple(result[2][0][1])[0] for result in results]
    supports    = [result[1] for result in results]
    return list(zip(lhs, rhs, supports))
resultsinDataFrame = pd.DataFrame(inspect(results), columns = ['Product 1', 'Product 2', 'Support'])
````

````python
resultsinDataFrame.nlargest(n = 10, columns = 'Support')
````
