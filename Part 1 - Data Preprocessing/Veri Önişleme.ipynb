# Veri önişleme toolları

## 1-Kütüphane aktarımı

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## 2-Veriseti Aktarımı

```python
dataset = pd.read_csv("Data.csv")
# iloc index locations demek sırası ile burada tüm satırlar dedik ilk başta sonrasında ilk üç sütun dedik x,y olarak düşün, 
# features yani özellikler verilerdeki burada ülke yaş maaş bu sebepten veriki 3 sütunu alıyoruz
x = dataset.iloc[:,:-1].values 
# depending variable yani tahmin etmeye çalıştığımı değer burada satın alınıp alınmadığı yes/no şeklinde son sütünu alıyoruz veride
y = dataset.iloc[:,-1].values #sadece son satırı alır
```

```python
print(x) #burada np array olarak oluşturdu unutma pd dataframe değil
```

```python
print(y) #son sütun
```

## 3-Eksik verilerin üstesinden gelmek

```python
from sklearn.impute import SimpleImputer
#Eksik Salary satır kısmına o sütun ortalama değerini eklemek
imputer = SimpleImputer(missing_values = np.nan, strategy='mean')
#oluşturduğumuz yöntemi ekliyoruz veriye
imputer.fit(x[:,1:3]) #yaş ve salary satırlarına ortalama değerleri ekleyecek
x[:,1:3] = imputer.transform(x[:,1:3]) #bu veriyi yeni verilerei ekliyor tam anlamı ile
```

```python
print(x)
```

## Kategorik verilerin kodlanması - Encoding categorical data

### Bağımsız değişlenlerin kodlanması- Encoding the Indepentent variable

```python
#Mevcut veri için konuşacak olursak makina öğrenimi 
#sayısal değer ister bu yüzden ülke isimlerini sayısala çeviricez
#burada unutma kategorik olan veriler için yapıyoruz
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
#ülke isimlerini encodluyoruz
ct = ColumnTransformer(transformers=[('encoder', OneHotEncoder(),[0])],remainder='passthrough')
x = np.array(ct.fit_transform(x)) #np array yapma sebebi birden fazla sütun var
```

```python
print(x) # dikkat edilirse ct de belirlediğimiz 0 sütununda olan ülk isimleri sadete encodlanacak
```

### Bağımlı değişkenlerin kodlanması - Encoding the dependent variable

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
y= le.fit_transform(y) #tek sütunluk bir veri olduğu için
```

```python
print(y) #satır etiketini 0 ve 1 olarak encodladık diyebiliriz
```

## Verisetini Test ve Eğitim Olarak iki parçaya ayırmak
## Splitting the dataset into the Training set and Test Set

```python
#en popüler olar sklearn train test split fonksiyonu veriyi bölmek için
from sklearn.model_selection import train_test_split
#random_state rastgele veri almasını sağlar 0.2 yüzde 20 olarak test ayırmasını sağlar
X_train, X_test, y_train, y_test = train_test_split(x,y,test_size= 0.2,random_state= 1)
```

```python
print(X_test)
```

## Özellik Ölçeklendirmesi
## Feature Scaling

✅ Neden fit_transform() sadece eğitim verisine uygulanır?
Çünkü:

fit() fonksiyonu ölçeklendirme (scaling) parametrelerini hesaplar:

    Örneğin StandardScaler için: ortalama (mean) ve standart sapma (std)

    MinMaxScaler için: min ve max değer

Eğer bu parametreleri hem eğitim hem test verisinden birlikte hesap edersen, test verisinin gelecekteki dağılım bilgisi modele sızar. Bu da:
❌ Data Leakage (Veri Sızıntısı) problemine yol açar.

Bu da modelin sahte bir başarıya ulaşmasına sebep olur, çünkü model test verisinin dağılımını önceden öğrenmiş olur.
🔄 Doğru yaklaşım:

    scaler.fit_transform(X_train)
    👉 Eğitim verisinden parametreleri öğren ve uygula

    scaler.transform(X_test)
    👉 Aynı parametreleri (mean, std, min, max vs.) test verisine uygula

Bu şekilde model, hiç görmediği veriye karşı doğru şekilde test edilir.

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
#standardscaleri sadece burada örnek vermek gerekirse kullanıcı yaşlarına maaşlarına uygulamak anlamlı olur
#çünkü onehot kodlaması yaptığımız ülkeler için saçma sonuçlar verir
#ülke kodlamalarını korumamız lazım burada
X_train[:,3:] = scaler.fit_transform(X_train[:,3:]) #ilk sütundaki değerlerin ortalamalarını alır sonra formüle uygular
X_test[:,3:] = scaler.transform(X_test[:,3:])
```
