## Genel tanım

Boston bölgesindeki konutlarla ilgili çeşitli sosyo-ekonomik ve çevresel değişkenleri içerir. Tipik kullanım, **ev değerini (MEDV)** tahmin eden bir regresyon modelidir. Veri kümesi klasik/öğretici bir benchmark olarak yaygınlaşmıştır. 

## Boyut ve dosyalar

* **Gözlem / özellik sayısı:** Sık kullanılan sürümde **506 gözlem** ve **14 sütun** (13 özellik + hedef) bulunur. Hedef değişken **MEDV**’dir.
* **Kaggle paket içeriği:** Bu sayfadaki sürüm tek bir **CSV dosyası** olarak paylaşılmıştır (yazarın veri kümeleri listesine göre). 

## Sütunlar (Türkçe açıklamalar)

* **CRIM:** Kişi başına düşen suç oranı (kasaba bazında)
* **ZN:** 25.000 ft² üzeri parseller için ayrılmış konut alanı oranı
* **INDUS:** Perakende dışı iş alanı oranı (kasaba bazında)
* **CHAS:** Charles Nehri kukla değişkeni (1 = nehire bitişik, 0 = değil)
* **NOX:** Azot oksit (NOx) konsantrasyonu (ppm)
* **RM:** Konut başına ortalama oda sayısı
* **AGE:** 1940’tan önce inşa edilmiş ve sahibi tarafından oturulan birimlerin oranı
* **DIS:** Boston’daki 5 istihdam merkezine ağırlıklı uzaklık
* **RAD:** Radyal otoyollara erişim endeksi
* **TAX:** 10.000\$ başına emlak vergi oranı
* **PTRATIO:** Öğrenci-öğretmen oranı (kasaba bazında)
* **B:** 1000·(Bk − 0.63)² (Bk: kasabadaki Black nüfus oranı)
* **LSTAT:** Düşük statülü nüfus yüzdesi
* **MEDV:** Sahip tarafından oturulan konutların ortanca değeri (bin \$) — **hedef**


## Etik / bağlamsal not

Bu veri kümesi, **ırk ve sınıf** ile ilişkili kalıpları (özellikle **B** ve **LSTAT** değişkenleri) içerebildiği için eğitimde adalet ve önyargı tartışmalarına konu olmuştur.
```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder,LabelEncoder
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("BostonHousing.csv")
```

```python
print("Missing values Perc. \n\n ", round(df.isnull().sum().sort_values(ascending=False)/len(df)*100,1))
#print(df.isna().any())
```

```python
X = df.iloc[:,:-1].values
y = df.iloc[:,-1].values
print("Len X", len(X), "Len y", len(y))
```

```python
imputer_rm = SimpleImputer(missing_values=np.nan, strategy='median')
rm_column = X[:,5].reshape(-1,1)
imputer_rm.fit(rm_column)
X[:,5] = imputer_rm.transform(rm_column).ravel()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state = 0)
```

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(X_train,y_train)
```

```python
y_pred = regressor.predict(X_test)
np.set_printoptions(precision=2)
print(
    np.concatenate(
        (
            y_pred.reshape(len(y_pred),1), 
            y_test.reshape(len(y_test),1)
        ),
        1)
)
```

```python
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error
r2 = r2_score(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
rmse = mean_squared_error(y_test, y_pred, squared=False)

print(f"Test R^2 : {r2:.3f}")
print(f"Test MAE: {mae:.3f}")
print(f"Test RMSE: {rmse:.3f}")
```
