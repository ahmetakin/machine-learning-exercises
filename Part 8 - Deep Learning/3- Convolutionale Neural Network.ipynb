## 🧠 **CNN Nedir?**

* **Yapay Sinir Ağlarının (ANN)** özel bir versiyonudur.
* Özellikle **görüntü işleme** görevleri için tasarlanmıştır.
* İnsan beyninin görsel korteksinden ilham alır.
* Resimlerdeki **uzamsal ilişkileri (spatial relationships)** yakalamada çok başarılıdır.

---

## 📦 **CNN’in Temel Katmanları**

Bu bölümde öğreneceğin katmanlar aslında CNN mimarisinin yapı taşları:

1. **Convolution (Evrişim) Katmanı**

   * Görüntü üzerinde **filtreler (kernels)** gezdirilerek **özellik çıkarımı** yapılır.
   * Örn: Kenarlar, köşeler, dokular yakalanır.
   * Çıktısı → **Feature Map** (Özellik haritası).

2. **ReLU (Rectified Linear Unit)**

   * Negatif değerleri 0 yapar, pozitifleri olduğu gibi geçirir.
   * **Lineer olmayan** dönüşüm ekler → karmaşık desenleri öğrenebilmek için gerekli.

3. **Pooling (Havuzlama)**

   * Görüntünün boyutunu küçültür (downsampling).
   * En yaygın: **Max Pooling** (küçük bir bölgedeki en büyük değeri alır).
   * Gürültüyü azaltır, işlem hızını artırır, önemli bilgileri korur.

4. **Flattening**

   * 2D feature map → 1D vektöre dönüştürülür.
   * Böylece **fully connected layer**’a bağlanabilir.

5. **Fully Connected Layer (Dense Layer)**

   * ANN mantığında çalışır.
   * Çıkarılan özellikleri birleştirir ve sınıflandırma yapar.

6. **Softmax + Cross-Entropy Loss**

   * Çok sınıflı sınıflandırmada her sınıf için olasılık verir.
   * Loss → cross-entropy ile hesaplanır.

## 📊 **CNN’in Çalışma Mantığı Görsel Olarak**

1. **Girdi:**

   * 64x64 piksel renkli resim → 3 kanal (RGB).
2. **Convolution + ReLU:**

   * 3x3 veya 5x5 filtreler → kenar, köşe, doku yakalanır.
3. **Pooling:**

   * 2x2 max pooling → boyut yarıya iner.
4. **Flattening:**

   * 2D matris → 1D vektör.
5. **Fully Connected:**

   * Özellikler birleştirilir, sınıflar belirlenir.
6. **Softmax:**

   * Her sınıfa ait olasılık hesaplanır.
7. **Output:**

   * “Köpek %95, Kedi %5” gibi.

## 🧠 1. **Convolution (Evrişim) Katmanı**

> 📌 Görevi: Görüntüden önemli **özellikleri** bulmak.

**Günlük hayat örneği:**
Bir fotoğrafa büyüteçle bakıp **sadece kenarlara** veya **sadece desenlere** dikkat etmek gibi.

📷 **Örnek:**

* Elinde bir kedi fotoğrafı var.
* Convolution katmanı küçük bir **filtre (ör. 3x3 pencere)** ile tüm fotoğrafı tarıyor.
* Bu filtre bazen yatay çizgilere duyarlı, bazen köşelere, bazen belirli renklere…
* Sonuç: "Kedinin kulağı burada, gözleri burada, bıyıkları burada" gibi **özellik haritaları** (feature maps) çıkar.

---

## ⚡ 2. **ReLU (Rectified Linear Unit)**

> 📌 Görevi: Negatif bilgileri sıfırlayarak **pozitif bilgiyi korumak**.

**Günlük hayat örneği:**
Bir sınav sonucunu değerlendirirken **eksi puanları yok saymak** gibi.

📷 **Örnek:**

* Filtre taramasından sonra bazı pikseller negatif değerli olabilir.
* ReLU der ki: “Negatifleri çöpe at, pozitifi koru.”
* Böylece önemli bilgiler kalır, gürültü azalır.

---

## 🌊 3. **Pooling (Havuzlama) Katmanı**

> 📌 Görevi: Görüntüyü **küçültmek**, ama **önemli bilgiyi kaybetmemek**.

**Günlük hayat örneği:**
Bir manzaraya uzaktan bakmak. Detaylar azalır ama genel şekil hâlâ bellidir.

📷 **Örnek:**

* 4x4 piksel bir alanın **en yüksek değerini** almak (Max Pooling).
* Bu sayede:

  * Görüntü boyutu yarıya iner.
  * Hesaplama yükü azalır.
  * Önemli özellikler korunur.

---

## 📜 4. **Flattening**

> 📌 Görevi: 2D görüntü bilgisini **düz bir listeye** çevirmek.

**Günlük hayat örneği:**
Bir Excel tablosundaki verileri satır satır kopyalayıp **tek bir liste haline getirmek** gibi.

📷 **Örnek:**

* Pooling sonrası 8x8 boyutunda özellik haritası kaldı diyelim.
* Flatten → `[0.2, 0.5, 0.9, ...]` şeklinde **tek boyutlu bir vektör** yapar.
* Çünkü sonraki katman (fully connected) **1D veri** ile çalışır.

---

## 🔗 5. **Fully Connected Layer (Dense Layer)**

> 📌 Görevi: Bütün özellikleri birleştirip **karar vermek**.

**Günlük hayat örneği:**
Birini tanımaya çalışırken yüz, saç rengi, ses tonu gibi tüm ipuçlarını bir araya getirip **“Bu Ahmet!”** demek gibi.

📷 **Örnek:**

* Flatten edilmiş özellikler buraya gelir.
* Ağ, “bu kulak şekli + bu bıyık uzunluğu + bu göz rengi → kedi” gibi karar kuralları öğrenir.
* Sonuçta tahmin üretir.

---

## 🎯 Son Adım: **Softmax + Cross-Entropy**

> 📌 Görevi: Çıktıyı olasılıklara dönüştürmek.

**Günlük hayat örneği:**
"Bu fotoğraf %95 ihtimalle kedi, %4 köpek, %1 tavşan" demek gibi.

---

## 🖼 **Görsel Düşün**

```
Girdi Resim → [Convolution] → [ReLU] → [Pooling] → [Flatten] → [Fully Connected] → Tahmin
```

Kısaca:

1. **Convolution:** Önemli noktaları bulur
2. **ReLU:** Negatifleri siler
3. **Pooling:** Boyutu küçültür
4. **Flatten:** Düz listeye çevirir
5. **Fully Connected:** Karar verir

### Importing the libraries

````python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator #resim işlemek için
````

````python
tf.__version__
````

## Part 1 - Data Preprocessing

### Preprocessing the Training set

````python
#overtrain e karşı görüntüleri sabitliyoruz geometrik transformasyon
# this is the augmentation configuration we will use for training
train_datagen = ImageDataGenerator(rescale = 1./255,
                                   shear_range = 0.2,
                                   zoom_range = 0.2,
                                   horizontal_flip = True)
# batches of augmented image data
training_set = train_datagen.flow_from_directory('dataset/training_set',  # this is the target directory
                                                 target_size = (64, 64),# all images will be resized to 64x64
                                                 batch_size = 32,
                                                 class_mode = 'binary') #since we use binary_crossentropy loss, we need binary labels
````

### Preprocessing the Test set

````python
# this is the augmentation configuration we will use for testing:
# only rescaling
test_datagen = ImageDataGenerator(rescale = 1./255)
test_set = test_datagen.flow_from_directory('dataset/test_set',
                                            target_size = (64, 64),
                                            batch_size = 32,
                                            class_mode = 'binary')
````

## Part 2 - Building the CNN

### Initialising the CNN

````python
cnn = tf.keras.models.Sequential()
````

### Step 1 - Convolution

````python
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu', input_shape=[64, 64, 3])) 
#32 filtreli 3d kernel relu activation input_Shape = 64X64X3 boyutlandırdığım boyut
````

### Step 2 - Pooling

````python
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
# kaç frame olarak alcak 2x2 o yüzden 2 pool_size strides ise shift kaç pixel ile olacağı
````

### Adding a second convolutional layer

````python
cnn.add(tf.keras.layers.Conv2D(filters=32, kernel_size=3, activation='relu'))
cnn.add(tf.keras.layers.MaxPool2D(pool_size=2, strides=2))
````

### Step 3 - Flattening

````python
cnn.add(tf.keras.layers.Flatten())
````

### Step 4 - Full Connection

````python
cnn.add(tf.keras.layers.Dense(units=128, activation='relu')) #yüksek tutuyoruz units i
````

### Step 5 - Output Layer

````python
cnn.add(tf.keras.layers.Dense(units=1, activation='sigmoid')) 
#ikili sınıflandırma yaptığımız için 1 nöron sigmoid ise sınıflandırma için uygun diye
````

## Part 3 - Training the CNN

### Compiling the CNN

````python
cnn.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
````

### Training the CNN on the Training set and evaluating it on the Test set

````python
cnn.fit(x = training_set, validation_data = test_set, epochs = 25)
````

## Part 4 - Making a single prediction

load_img: Resmi yükler.

target_size: Eğitimdeki boyutla aynı hale getirir (64x64).

img_to_array: NumPy dizisine çevirir.

expand_dims: Tek resim olduğu için batch boyutunu ekler.

predict: Olasılık tahmini yapar.

class_indices: Modelin hangi etiketi hangi sayıya atadığını gösterir.

````python
import numpy as np
from tensorflow.keras.preprocessing import image
test_image = image.load_img('dataset/single_prediction/cat_or_dog_1.jpg', target_size = (64, 64)) #resimleri 64 64 yapmıştık o yüzden bunuda yapıyoruz
test_image = image.img_to_array(test_image)  # Resmi NumPy array'e çevir
test_image = np.expand_dims(test_image, axis = 0) # (1, 64, 64, 3) şekline getir
result = cnn.predict(test_image)
training_set.class_indices # {'cats': 0, 'dogs': 1}
if result[0][0] == 1:
  prediction = 'dog'
else:
  prediction = 'cat'
````
